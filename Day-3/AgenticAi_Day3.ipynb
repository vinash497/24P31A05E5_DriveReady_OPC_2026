```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Initialize model
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.3
)

# Create prompt
prompt = ChatPromptTemplate.from_messages([
    ("system", SYSTEM),
    ("human", "{question}")
])

# Create chain
chain = prompt | llm | StrOutputParser()

print("GreenNest chain created successfully!")
```

```python
#GreenNest Business Selection and System Prompt
SYSTEM = """
You are GreenNest's friendly AI customer support assistant.

ROLE:
You help customers choose indoor plants, understand plant-care
requirements, check product prices, and enquire about orders.

BUSINESS FACTS:
- Snake Plant: Rs. 450
- Money Plant: Rs. 250
- Peace Lily: Rs. 550
- Areca Palm: Rs. 700
- ZZ Plant: Rs. 600
- Decorative pot: Additional Rs. 100
- Every plant includes a basic care instruction card.
- Orders require at least 12 hours' notice.
- Delivery is available only within the business service area.

BUSINESS RULES:
1. Never invent product prices or business information.
2. Check product availability before confirming an order.
3. Never claim an order is confirmed without actual confirmation.
4. If a customer asks about an unknown delivery location,
   respond: "I'll check with our delivery team."
5. If you do not know an answer, never guess.
6. Calculate product totals carefully.
7. Provide simple and practical plant-care suggestions.
8. Never guarantee that a plant will survive in every environment.

RESPONSE FORMAT:
- Be friendly, polite, and helpful.
- Keep answers within 60 words.
- Use simple English.
- Answer the customer's question directly.
- Use bullet points when explaining multiple details.
- Ask follow-up questions when essential order information is missing.
"""
```

```python

import csv

def run_tests(path="test.csv"):
    passed = 0
    total = 0

    with open(path, "r", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        for row in reader:
            question = row["question"]
            expected = row["must_include"]

            response = chain.invoke({
                "question": question,
                "history": []
            })

            total += 1

            if expected.lower() in response.lower():
                passed += 1
                status = "PASS"
            else:
                status = "FAIL"

            print("Question:", question)
            print("Expected:", expected)
            print("Response:", response)
            print("Status:", status)
            print("-" * 50)

    pass_rate = (passed / total) * 100

    print("Total Tests:", total)
    print("Passed:", passed)
    print("Failed:", total - passed)
    print("Pass Rate:", pass_rate, "%")

    return pass_rate

run_tests("test.csv")
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[check_stock, calculate_price],
    system_prompt="""
    You are GreenNest's ordering assistant.

    1. Understand the customer's order.
    2. Check plant availability using check_stock.
    3. Calculate the total using calculate_price.
    4. Explain the result clearly.
    5. Never claim an order is confirmed.
    """
)

print("GreenNest AI Agent created successfully!")
```

```python
from langchain_core.tools import tool

@tool
def check_stock(plant: str) -> str:
    """Check availability of a GreenNest plant."""

    stock = {
        "snake plant": 10,
        "money plant": 15,
        "peace lily": 5,
        "areca palm": 8,
        "zz plant": 7
    }

    quantity = stock.get(plant.lower())

    if quantity is None:
        return "Plant not found in catalogue."

    return f"{quantity} units available."


@tool
def calculate_price(plant: str, quantity: int,
                    pots: int = 0) -> str:
    """Calculate the total order price."""

    prices = {
        "snake plant": 450,
        "money plant": 250,
        "peace lily": 550,
        "areca palm": 700,
        "zz plant": 600
    }

    price = prices.get(plant.lower())

    if price is None:
        return "Plant not found in catalogue."

    total = (price * quantity) + (100 * pots)

    return f"Total order price: Rs. {total}"
```

```python
result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "I want 2 Peace Lilies with 2 decorative pots. Check availability and calculate the total price."
        }
    ]
})

print(result["messages"][-1].content)
```
